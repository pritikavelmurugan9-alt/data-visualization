```python
import matplotlib.pyplot as plt
x = [1,3,6,7,9]
y = [55, 60,79,88,99]
x1=[1,2,3,4,5]
y1=[59,65,72,85,100]
plt.xlabel("ROLL NUMBER")
plt.ylabel("SUBJECT MARKS")
plt.xlim(1, 10)
plt.ylim(50, 100)
plt.title("STUDENT PERFORMANCE")
plt.plot(x,y,label="Line-1",color='grey',linestyle='dashdot',linewidth=4,marker='*',markerfacecolor='blue',markersize=13) 
plt.plot(x1,y1,label="Line-2",alpha=0.9,color='grey',linestyle='dashed',linewidth=4,marker='D',markerfacecolor='red',markersize=11) 
plt.legend()
plt.savefig("Sample1.png")
plt.show()
```
