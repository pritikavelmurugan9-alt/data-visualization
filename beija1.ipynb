```python
import pandas as pd
```

```python
df=pd.read_csv("C:/Users/Ibscai37.DGVC-MCA/Downloads/beija.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.fillna(0)
```

```python
mn=df['High'].mean()
mn
```

```python
df['High']=df['High'].fillna(mn)
df.isnull().sum()
```

```python
md=df['Open'].median()
md
```

```python
df['Open']=df['Open'].fillna(md)
df.isnull().sum()
```

```python
ms=df['Low'].mode()
ms
```

```python
df['Close']=df['Close'].bfill()
df.isnull().sum()
```

```python
df['Low']=df['Low'].bfill()
df.isnull().sum()
```

```python
df['Volume']=df['Volume'].bfill()
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
x=df['Open']
y=df['High']
plt.title("BEIJA")
plt.plot(x,y,label="Line-1",color='grey',linestyle='dashdot',linewidth=4,marker='*',markerfacecolor='blue',markersize=13) 
plt.plot(x,y)
plt.show()
```
